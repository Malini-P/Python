## Rock,Paper,Scissors Game

Objective:
Build a simple Rock-Paper-Scissors game where the player competes with the computer.
 This task will test your understanding of:
●	Input/output
●	Conditional logic
●	Random module
●	Loops
●	Exception handling

Instructions:
1.	The user should enter their choice: rock, paper, or scissors.
2.	The computer should randomly select one of these.
3.	Compare both and print who wins.
4.	Ask if the player wants to play again- if yes, repeat; if no, exit.
5.	Handle invalid inputs gracefully using try-except or condition checks.

```python
def hand_game():

    import random
    
    print("Welcome to Rock,Paper,Scissor Game !!!")
    print("==============================================")
    
    clue = ["rock", "paper" , "scissor"]
   
    while True:
        
        computer_choice = random.choice(clue)
        user_choice = input("Enter your choice:")
        
        if user_choice not in clue:
            print("Invalid Input. Please enter the correct clue")
            continue
            
        print("The computer choice is: {}".format(computer_choice))
        
        if user_choice == computer_choice:
            print("Match draw🤷‍♀️")
        elif (user_choice == "rock" and computer_choice == "scissor") or (user_choice == "paper" and computer_choice == "rock") or (user_choice == "scissor" and computer_choice == "paper"):
            print("User won🎉")
        else:
            print("User loss")
            
        play_again = input("Do you want to play again?")
        if play_again not in ["Yes" , "yes", "y"]:
            print("Thanks for playing... GoodBye👋")
            break
```

```python
hand_game()
```

## Word Guessing Game (Mini Hangman)
Objective:
Create a word-guessing game where the player tries to guess the letters of a hidden word.

This will test your:
●	Loops
●	Conditional logic
●	String manipulation
●	Exception handling
●	Logical thinking

Instructions:
1.	Create a list of random words (like “python”, “machine”, “learning”, “bridge”).
2.	Randomly select one word for the game.
3.	Show the word as blanks (_ _ _ _) and let the player guess one letter at a time.
4.	Keep track of wrong guesses (max 6 attempts).
5.	End the game when the player either guesses the full word or uses all chances.

```python
def mini_hangman():
    
    import random
    
    words = ["python","machine","learning","bridge"]
    word = random.choice(words)
    
    guessed = ["_"] * len(word)
    attempts = 6
    
    print("Welcome to the Mini Hangman")
    print("===============================")
    print("Guess the word")
    
    while attempts > 0:
        print("The clue:"," ".join(guessed))
        letter = input("Enter a letter:")
    
        if letter in word:
            new_word = ""
            for i in range(len(word)):
                if word[i] == letter:
                    new_word = new_word + letter
                else:
                    new_word = new_word + guessed[i]
            guessed = new_word
        else:
            attempts = attempts - 1
            print("You are wrong !!! Attempts left:" ,attempts)
    
        if guessed == word:
            print("You won 🎉 Yes!!! The word is:",word)
            break
    
    if guessed != word:
        print("You lost 😒 The word is:",word)
```

```python
mini_hangman()
```
