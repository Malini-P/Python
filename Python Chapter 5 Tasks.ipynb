### Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Import Dataset

```python
sns.get_dataset_names()
```

```python
titanic_data = sns.load_dataset(name = 'titanic' )
titanic_data
```

```python
titanic_data.shape
```

```python
titanic_data.dtypes
```

```python
titanic_data.isna().sum()
```

```python
titanic_data.describe(include = 'all')
```

### 2. Data cleaning

```python
titanic_cleaned_data = titanic_data.dropna(subset = ["age" , "deck", "embark_town" , "embarked"])
titanic_cleaned_data
```

```python
titanic_cleaned_data.dtypes
```

```python
titanic_cleaned_data.isna().sum()
```

### 3. Data Visualization

```python
titanic_cleaned_data.head()
```

## 3.1 Univariate Analysis

## Histogram

```python
plt.hist(titanic_cleaned_data["age"])
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
```

```python
sns.histplot(data=titanic_cleaned_data , x="fare" ,kde=True, bins=20, color="Green" ) #kde = kernel density estimate
plt.xlabel("Fare")
plt.title("Fare Distribution with KDE")
plt.show()
```

```python
sns.histplot(data=titanic_cleaned_data, x='age' , hue='alive', multiple='stack')
plt.xlabel("Age")
plt.title("Age vs Alive distribution")
plt.show()
```

## Countplot

```python
sns.countplot(data = titanic_cleaned_data, x = "embark_town"  , hue = "alive")
plt.xlabel("Embark Town")
plt.title("Embark Town vs Gender Distribution")
plt.show()
```

## PieChart

```python
titanic_cleaned_data["who"].value_counts()
```

```python
plt.pie(x=titanic_cleaned_data["who"].value_counts(), labels = ["Man", "Woman" , "Child"] , explode = [0.01,0.01,0.01] , autopct = "%1.0f%%")
plt.title("Total Gender Distribution in Titanic Ship")
plt.show()
```

```python
titanic_cleaned_data["alive"].value_counts()
```

```python
plt.pie(x=titanic_cleaned_data["alive"].value_counts(), labels = ["Survived", "Dead"] , explode = [0.01,0.01] , autopct = "%1.0f%%")
plt.title("Alive Distribution in titanic Ship")
plt.show()
```

```python
titanic_cleaned_data["deck"].value_counts()
```

```python
plt.pie(x=titanic_cleaned_data["deck"].value_counts(), labels = ["C","B","D","E","A","F","G"] , explode = [0.01,0.01,0.01,0.01,0.01,0.01,0.01] , autopct = "%1.0f%%")
plt.title("Passengers % in Each Deck Distribution")
plt.show()
```

## 3.2 Bivariate Analysis

## BarPlot

```python
sns.barplot(data=titanic_cleaned_data, x="pclass", y="survived")
plt.xlabel("PClass")
plt.ylabel("Survived")
plt.title("PClass vs Survivors Distribution")
plt.show()
```

```python
sns.barplot(data=titanic_cleaned_data, x="who", y="age", hue="class")
plt.xlabel("Passenger Categorization")
plt.ylabel("Age")
plt.title("Passenger Categorization based on Age & Class")
plt.show()
```

## ScatterPlot

```python
plt.scatter(x="fare", y="age", data=titanic_cleaned_data)
plt.xlabel("Fare")
plt.ylabel("Age")
plt.title("Age vs Fare Distribution")
plt.show()
```

## 3.3 Multivariate Analysis

## PairPlot

```python
sns.pairplot(data = titanic_cleaned_data)
plt.show()
```

## HeatMap

```python
sns.heatmap(titanic_cleaned_data.corr(numeric_only=True), annot=True)
plt.title("Heatmap for Titanic Dataset")
plt.show()
```
