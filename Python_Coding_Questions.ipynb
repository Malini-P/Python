## 2.Create a list of numbers and print only even numbers using a for loop.

```python
numbers = range(1,51)
even_numbers = []

for i in numbers:
    if i%2 == 0:
        even_numbers.append(i)
        
print(even_numbers)    
```

## 3.Write a program that handles division by zero using try-except.

```python
try:
    num = int(input("Enter Numerator:"))
    den = int(input("Enter Denominator:"))

    result = num/den
    print(result)
except ZeroDivisionError:
        print("Invalid Entry! Enter a Number other than 0")
except ValueError:
        print("Invalid Entry! Enter a Number not character")
```

## 4.Given a string, count how many vowels it has.

```python
user_input = input("Enter a word or sentence:")
count = 0

for i in user_input.lower():
    if i in "aeiou":
        count+=1

print("Total number of vowels:",count)
```

## 5.Reverse a list without using built-in functions.

```python
original_list = [1,2,3,4,5]
original_list.reverse()
original_list
```

```python
original_list = [1,2,3,4,5]
reverse_list = []
i = len(original_list) - 1

while i>=0:
    reverse_list.append(original_list[i])
    i = i-1

print(reverse_list)
```
