## Libraries

It is a collection of pre-written code (modules, functions and classes) that you can use to perform common tasks without writing everything from scratch.

* Module - ex: math_utils
* Function - ex: greet
* Class

#### Why we use libraries?

* Save time - no need to write code for common tasks.
* Proven solutions - tested and optimized by experts.
* Easy to use - just import and use the functions.

#### Types of Libraries

* Pandas                        -  Data Analysis
* Numpy                         -  Math/Science
* matplotlib , Seaborn          -  Data Visualization
* scikit-learn                  -  Machine learning algorithms
* Tensorflow , Keras , Pytorch  -  Deep Learning
* Flask , django                -  Web Development

## 2. Numpy

```python
import numpy as np
```

```python
np.repeat(a = sales_data , repeats = 2, axis = 0) 
# It repeats the datas into 2 times , axis = 0 mean it represents rows if it is 1 means it represents column
```

```python
import plotly.express as px
df = px.data.gapminder().query("year == 2007").query("continent == 'Europe'")
df.loc[df['pop'] < 2.e6, 'country'] = 'Other countries' # Represent only large countries
fig = px.pie(df, values='pop', names='country', title='Population of European continent')
fig.show()
```
