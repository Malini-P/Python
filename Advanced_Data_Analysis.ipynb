## 1.Groupby Function

```python
import pandas as pd

insurance_data = pd.read_csv("insurance.csv")
insurance_data
```

### 1.1 Perform Initial Analysis

```python
insurance_data.isna().sum()
```

```python
insurance_data.dtypes
```

```python
insurance_data.describe(include = "all")
```

```python
insurance_data["region"].unique()
```

### 1.2 Pull up the Insurance Charges based on Region

```python
insurance_data.groupby(by="region")["charges"].mean().sort_values().round(2)
```

```python
insurance_data.groupby(by="region")["charges"].mean().sort_values(ascending=False).round(2)
```

### 1.2 Pull up the average Insurance charges based on region and sex

#### Discrete / categorical data and Continous data

* Discrete - Just countable / Can't be segmented
* Continous - Measurable / Have Units / can be segmented

### NOTE : 

### In this groupby function(by -> it should contains one or more but the values must be discrete) [" " -> Outside it should contain only continous]

```python
insurance_data.head(7)
```

```python
insurance_data.groupby(by=["region","sex"])["charges"].mean().sort_values(ascending=False).round(2)
```

```python
insurance_data.groupby(by=["region","sex"])["charges"].max().sort_values(ascending=False).round(2)
```

## 2.Pivot Table

```python
pd.pivot_table( data=insurance_data, values="charges", index="region", columns="sex").round() #presentable comparing 2 categorical features vs  1 continous data
```

```python
pd.pivot_table( data=insurance_data, values="charges", index="region", columns=["sex","children"]).round()
```

## 3.Aggregate

```python
insurance_data.head()
```

### 3.1 Find out the average Insurance charges based on Region

```python
insurance_data.groupby(by="region")["charges"].mean().sort_values().round(2)
```

```python
insurance_data.groupby(by=["region","sex"])["charges"].agg(func = ["mean","min","max","sum"]).round()
```

#### 6.1 Create a new column in the same dataframe with the name "Eligibility Criteria" where if age is is > 20, say Eligible else not

```python
insurance_data["age"] > 20
```

```python
if insurance_data["age"] > 20:
    print("Eligible")
else:
    print("Not Eligible")
```

```python
def get_age(x):
    if x > 20:
          print("Eligible")
    else:
          print("Not Eligible")
```

```python
insurance_data["Eligibility criteria"] = insurance_data["age"].apply( get_age)
insurance_data
```

```python
def get_age(x):
    if x > 20:
         return "Eligible"
    else:
         return "Not Eligible"
```

```python
insurance_data["Eligibility criteria"] = insurance_data["age"].apply( get_age)
insurance_data
```

## 7.Lambda

It is also called as Anonymous function or Nameless function

```python
insurance_data
```

```python
insurance_data["age"].apply(lambda x:"Eligible" if x>20 else "Not eligible")
```

```python
insurance_data["EC"] = insurance_data["age"].apply(lambda x:"Eligible" if x>20 else "Not eligible")
insurance_data
```
