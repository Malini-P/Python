## Real - world analogy 1:

### I want to reuse the same function for Team Members and Managers, but they have small differences. How do I do that?

Both Team Members and Managers should be able to log in, but:

   * Team Members - should log in, view tasks, and submit tasks.
   * Managers - should log in, view tasks, and assign tasks.

```python
#Defining a function
def login(user_data):
    print(f"{user_data['Name']} has logged in")
    
def view_tasks(user_data):
    print(f"{user_data['Name']} has viewing the task")
    
def submit_tasks(user_data):
    if user_data["Role"] == "Team_Member":
        print(f"{user_data['Name']} has submitted the task")
    else:
        print("Managers no need to submit the task")
    
def assign_tasks(user_data):
    if user_data["Role"] == "Manager":
        print(f"{user_data['Name']} has assigned the task")
    else:
        print("Team members can't assigned the task")




#Data Collection
user_data = {"Name":"Malini" , "Role":"Manager"}
#user_data = {"Name":"Yujin", "Role":"Team_Member"}

#Functions to be called
login(user_data)
view_tasks(user_data)
submit_tasks(user_data)
assign_tasks(user_data)
```

### Now, we can understand the limitation of Function - Module coding

### With OOPS

* Class - Blueprint that contains attributes of a particular objects and it's all the functions
* Object - Instance of a class

```python
class Employee:
    def __init__(self,name):
        self.name = name
        
    def login(self):
       print(f"{self.name} has logged in")
    
    def view_tasks(self):
       print(f"{self.name} has viewing the task")

class TeamMember(Employee):
    def submit_tasks(self):
        print(f"{self.name} has submitted the task")

class Manager(Employee):
    def assign_tasks(self):
        print(f"{self.name} has assigned the task")
```

```python
john = TeamMember("John")
nasir = Manager("Nasir")
```

```python
john.login()
john.submit_tasks()
john.view_tasks()
```

```python
nasir.assign_tasks()
nasir.login()
nasir.view_tasks()
```
