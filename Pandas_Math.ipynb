## Module file

```python
def addition(a,b):
    return a+b

def subtraction(a,b):
    return a-b

def multiply(a,b):
    return a*b

def divide(x,y):
    return x/y
```

## Terminology Alert

* Parameters - While creating the function, you create parameters
* Attributes - While calling the function, you pass arguments

## Difference between Return and Print

* Return - Impacts the original Data when the function is called
* Print - It won't impact the original Data, but just prints
