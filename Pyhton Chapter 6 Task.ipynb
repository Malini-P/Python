## Encapsulation

* Encapsulation is the process of wrapping data and methods into a single unit (class) and restricting direct access to some of the data. It helps in protecting sensitive information.

* In this ATM example, the atm_pin is declared as a private variable (__atm_pin), so it cannot be accessed directly from outside the class. Instead, it is accessed through a method like get_pin().

* This ensures data security and controlled access to important information like the user’s PIN.

```python
class SecureUser:
    
    def __init__(self, name, atm_pin):
        self.name = name
        self.__atm_pin = atm_pin
        self.balance = 0

    def get_pin(self, attempt):
        if attempt == self.__atm_pin:
            return True
        else:
            return False

    def deposit_cash(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def cash_withdrawal(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance")
```

```python
user_data = SecureUser("Yujin", 2511)
```

```python
user_data.get_pin(2344)
```

```python
user_data.deposit_cash(500)
```

```python
user_data.cash_withdrawal(100)
```

```python
user_data.balance
```

## Inheritance

* Inheritance is a concept in Object-Oriented Programming where one class (child class) acquires the properties and methods of another class (parent class). It helps in code reusability and reduces duplication.

* In this ATM example, the ATMUser class inherits from the SecureUser class. This means ATMUser can use features like PIN verification, deposit, and withdrawal without rewriting the code.

* This makes the program more organized and easier to maintain, as common functionalities are defined in a single parent class and reused by child classes.

```python
class ATMUser(SecureUser):

    def show_balance(self):
        print("Balance:", self.balance)
```

## Abstraction

* Abstraction is the process of hiding internal implementation details and showing only the essential features to the user. It helps in reducing complexity and improves code clarity.

* In this ATM example, an abstract class defines methods like deposit_cash and cash_withdrawal without providing their full implementation. The child classes implement these methods based on their specific behavior.

* This ensures that the user interacts only with the necessary functionalities, without knowing how they are internally implemented.

```python
#We hide implementation using base class


from abc import ABC, abstractmethod

class ATMOperations(ABC):

    @abstractmethod
    def deposit_cash(self, amount):
        pass

    @abstractmethod
    def cash_withdrawal(self, amount):
        pass
```

```python
#Now connect it with your class:


class ATMUser(SecureUser, ATMOperations):

    def show_balance(self):
        print("Balance:", self.balance)
```

```python
user_data_2 = ATMUser("Yujin", 2511)
```

```python
user_data_2.show_balance()
```

```python
user_data_2.deposit_cash(900)
```

```python
user_data_2.cash_withdrawal(700)
```

```python
user_data_2.show_balance()
```

## Polymorphism

* Polymorphism means “many forms” — it allows the same method name to behave differently based on the object that calls it. It improves flexibility and scalability in a program.

* In this ATM example, different classes like SBIUser and HDFCUser use the same method cash_withdrawal(), but each provides its own implementation.

* This allows the program to perform similar operations in different ways, depending on the type of user or bank.

```python
class SBIUser(ATMUser):

    def cash_withdrawal(self, amount):
        print("SBI ATM Processing...")
        super().cash_withdrawal(amount)


class HDFCUser(ATMUser):

    def cash_withdrawal(self, amount):
        print("HDFC ATM Processing...")
        super().cash_withdrawal(amount)
```

```python
user1 = SBIUser("Malini", 1234)

if user1.get_pin(1234):
    user1.deposit_cash(5000)
    user1.cash_withdrawal(2000)
    user1.show_balance()
else:
    print("Wrong PIN")
```
