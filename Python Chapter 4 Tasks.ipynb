### 2. Data Cleaning

#### Step 1: Drop the unwanted columns

```python
sales_full_data = sales_full_data.drop(columns = ["Gross","Disc","Voucher Amount"])
sales_full_data
```

#### Step 2: Remove the null values in every column

```python
sales_full_data = sales_full_data.dropna(subset = ["Date","Voucher","Party","Product","Qty","Rate"])
sales_full_data
```

#### Step 3: Check the Total observations and features

```python
sales_full_data.shape
```

#### Step 4: Modifications in Voucher column

```python
sales_full_data["Voucher"] = sales_full_data["Voucher"].astype(str).str.replace('Sal:','').str.strip() #Converting the Voucher Object into string and replace the Sal: into empty spaces also strip the entire value
sales_full_data = sales_full_data[sales_full_data["Voucher"] != '']  #If Voucher is not equal to empty spaces it should sliced and updated in the dataset.
sales_full_data["Voucher"] = sales_full_data["Voucher"].astype(int)  #The Voucher should converted from string to integer
```

```python
sales_full_data
```

```python
sales_full_data.dtypes
```

```python
sales_full_data["Voucher"].isna().sum()
```

#### Step 5: Modifications in Rate column

```python
sales_full_data["Rate"] = sales_full_data["Rate"].astype(str).str.replace(',','') #Converting the rate from object to string datatype and replace the , into no space
sales_full_data["Rate"] = pd.to_numeric(sales_full_data["Rate"], errors='coerce') #Convert the string into numeric
```

```python
sales_full_data
```

```python
sales_full_data.dtypes
```

```python
sales_full_data["Rate"].isna().sum()
```

#### Step 6: Modifications in Qty column

```python
sales_full_data["Qty"] = sales_full_data["Qty"].astype(str).str.replace(',','') #Convert object into string and replace (,) into empty space
sales_full_data["Qty"] = pd.to_numeric(sales_full_data["Qty"])
sales_full_data["Qty"] = sales_full_data["Qty"].astype(int)                     #Convert string into int
```

```python
sales_full_data
```

```python
sales_full_data.dtypes
```

```python
sales_full_data["Qty"].isna().sum()
```

#### Step 7: Reset the index

```python
sales_full_data = sales_full_data.reset_index(drop=True)
sales_full_data
```

#### Step 8: Change the Date column from object to Datetime dtype

```python
sales_full_data["Date"] = pd.to_datetime(sales_full_data["Date"] , dayfirst=True)
sales_full_data
```

## Merge Function

```python
import pandas as pd
```

```python
orders_data = pd.read_csv("orders.csv")
orders_data
```

```python
customers_data = pd.read_csv("customers.csv")
customers_data
```

### Inner join

Keeps only customers that exist in both datasets
* co1,co2,co3,co4 matched
* co5,co6 dropped(not in customers)
* co7,co8 dropped(not in orders)
* co1 and co3 appear twice because they placed 2 orders each

Result 6 rows

```python
inner_join = pd.merge(orders_data,customers_data, on="Customer_ID" , how="inner")
inner_join
```

### Left join 

All of left(orders) match from right(customers)

* Keeps all 8 rows from orders(left table) no matter what
* Matches customer info where available
* co5 -> no match in customers -> Name,City,Segment = NaN
* co6 -> no match in customers -> Name,City,Segment = NaN
* co7, co8 dropped (We can only care about orders here)

Result 8 rows

```python
left_join = pd.merge(orders_data,customers_data, on="Customer_ID" , how="left")
left_join
```

### Right join

All of right(Customers), match from eft(orders)

* Keeps all 6 rows from customers(right table) no matter what
* Matches order info where available
* co7 -> no orders palced -> Order_ID,Product,Qty = NaN
* co8 -> no orders palced -> Order_ID,Product,Qty = NaN
* co5, co6 dropped(not in customers)

Result 8 rows


```python
right_join = pd.merge(orders_data,customers_data, on=None, how="right")
right_join
```

### Outer join

Everything from both tables

* Keeps all rows from both tables
* co1,co2,co3,co4 -> Matched full info
* co5,co6 -> only in orders -> Customer info = NaN
* co7,co8 -> only in customers -> order info = NaN

Result 10 rows (largest of all joins)

```python
outer_join = pd.merge(orders_data,customers_data, on=None, how="outer")
outer_join
```

## Lambda Function

```python
import pandas as pd
```

```python
orders = pd.read_csv("orders.csv")
orders
```

```python
customers = pd.read_csv("customers.csv")
customers
```

### Lambda Function using apply function

```python
orders["Order_Type"] = orders["Qty"].apply(lambda x: "Bulk" if x>= 3 else "regular")
orders[["Product" , "Qty", "Order_Type"]]
```

```python
orders
```

### Lambda function using map in apply to the dataset

* Map applies a function to every element in a list and returns a new list of the same size.

Syntax:

list(map(lambda x: expression, list))

#### Ex 1:

```python
registered_ids = list(customers["Customer_ID"])
orders["Is_Registered"] = list(map(lambda x: "Registered" if x in registered_ids else "Unregistered", orders["Customer_ID"]))
orders[["Customer_ID", "Is_Registered"]]
```

```python
orders
```

#### Ex 2:

```python
metro_cities = ["Chennai", "Mumbai", "Delhi" , "Bangalore"]
customers["City_Type"] = list(map(lambda x: "Metro" if x in metro_cities else "Non-Metro" , customers["City"]))
customers["Priority"] = customers["City_Type"].apply(lambda x: "High priority" if x == "Metro" else "Normal Priority")
customers[["City","City_Type" ,"Priority"]]                                
```

```python
customers
```

### Lambda function using Filter in apply to the dataset

* filter picks only the elements that satisfy a condition and returns a smaller list

Syntax

list(filter(lambda x: condition, list))

#### Ex 1:

```python
order_cust_list = list(orders["Customer_ID"])
customers["Order_Count"] = customers["Customer_ID"].apply(lambda x: len(list(filter(lambda o:o == x, order_cust_list))))
customers[["Name","Order_Count"]]
```

```python
customers
```

#### Ex 2:

```python
customers["Has_Order"] = customers["Customer_ID"].apply(lambda x: "Yes" if list(filter(lambda o:o == x, order_cust_list)) else "No")
customers[["Name","Has_Order"]]
```

```python
customers
```

### Lambda using Reduce function in apply to the dataset

* Reduce combines all elements into a single value by applying a function repeatedly.

Syntax:

from functools import reduce
reduce(lambda a , b: expression, list)

#### Ex 1:

```python
import pandas as pd
from functools import reduce
```

```python
qty_cust_list = list(zip(orders["Customer_ID"], orders["Qty"]))
customers["Total_Qty"] = customers["Customer_ID"].apply(lambda x: reduce(lambda a, b: a+b, 
                                        [q for c, q in qty_cust_list if c == x])
                                    if [q for c , q in qty_cust_list if c==x] else 0)
customers[["Name","Total_Qty"]]
```

```python
customers
```

#### Ex 2:

```python
customers["Max_Qty"] = customers["Customer_ID"].apply(lambda x: reduce(lambda a, b: a if a> b else b, 
                                                [q for c, q in qty_cust_list if c == x])
                                            if[q for c, q in qty_cust_list if c==x] else 0)
customers[["Name","Max_Qty"]]
```
