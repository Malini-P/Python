### 2. Sets
* Set is One dimensional (1D)
* It is heterogeneous
* Set won't allow duplicate values
* Indexing and Slicing is not possible
* Set is mutable

```python
home_appliances = set()
type(home_appliances)
```

```python
home_appliances = {"Chair","Table","TV","AC","Cot","Chair"}  #No duplicate values
home_appliances
```

#### 1. add - Add an element to a set.

This has no effect if the element is already present.

```python
home_appliances = {"Chair","Table","TV","AC","Cot"}  #Set always arrange the element in ascending order like alphabets and numerics
home_appliances.add("Fridge")
home_appliances
```

#### 2. clear - Remove all elements from this set.

```python
home_appliances = {"Chair","Table","TV","AC","Cot"}
home_appliances.clear()
home_appliances
```

#### 3. copy - Return a shallow copy of a set.

```python
home_appliances = {"Chair","Table","TV","AC","Cot"}
home_appliances.copy()
home_appliances
```

#### 4. difference - Return the difference of two or more sets as a new set.

(i.e. all elements that are in this set but not the others.)

```python
home_appliances = {"Chair","Table","TV","AC","Cot"}
home_appliances_1 = {"Chair","Table","TV","AC"}
print(home_appliances.difference(home_appliances_1))
#home_appliances
```

#### 5. difference_update -  Remove common elements from the original set(it updates the set itself)

```python
home_appliances = {"Chair","Table","TV","AC","Cot"}
home_appliances_1 = {"Chair","Table"}
home_appliances.difference_update(home_appliances_1)
print(home_appliances)
```

#### 6. discard - Remove an element from a set if it is a member.

If the element is not a member, do nothing.

```python
home_appliances = {"Chair","Table","TV","AC","Cot"}
home_appliances.discard("AC")
home_appliances
```

```python
home_appliances = {"Chair","Table","TV","AC","Cot"}
home_appliances.discard("Fridge")
home_appliances
```

#### 7. intersection - Return the intersection of two sets as a new set.

(i.e. all elements that are in both sets.)

```python
home_appliances = {"Chair","Table","TV","AC","Cot"}
home_appliances_1 = {"Chair","Table","Frige","Sofa","TV"}
home_appliances.intersection(home_appliances_1)
#home_appliances
```

#### 8. intersection_update - Update a set with the intersection of itself and another.

```python
home_appliances = {"Chair","Table","TV","AC","Cot"}
home_appliances_1 = {"Chair","Table","Frige","Sofa","TV"}
home_appliances.intersection_update(home_appliances_1)
print(home_appliances)
home_appliances_1.intersection_update(home_appliances)
print(home_appliances_1)
```

#### 9. isdisjoint - Return True if two sets have a no common elements

```python
home_appliances = {"Chair","Table","TV","AC","Cot"}
home_appliances_1 = {"Chair","Table","Frige","Sofa","TV"}
home_appliances.isdisjoint(home_appliances_1)
```

```python
home_appliances = {"Chair","Table","TV","AC","Cot"}
home_appliances_1 = {"Sofa","Washing Machine","Fan","Mixer",45, 34, 23}
home_appliances.isdisjoint(home_appliances_1)
```

#### 10. issubset - Report whether another set contains this set.
Check Is set1 inside set2 ?
* Smaller -> Bigger

```python
home_appliances = {"Chair","Table","TV","AC","Cot"}
home_appliances_1 = {"Sofa","Washing Machine","Fan","Mixer",45, 34, 23}
home_appliances.issubset(home_appliances_1)
```

```python
home_appliances = {"Chair","Table","TV","AC","Cot"}   #Here the total home_appliances set is completely present in the second set
home_appliances_1 = {"Chair","Table","Frige","Sofa","TV","AC","Cot"}
home_appliances.issubset(home_appliances_1)
```

#### 11. issuperset - Report whether this set contains another set
Check Does set1 contain set2?
* Bigger -> Smaller

```python
home_appliances = {"Chair","Table","TV","AC","Cot"}
home_appliances_1 = {"Fridge","sofa","Table","TV"}
home_appliances.issuperset(home_appliances_1)
```

```python
home_appliances = {"Chair","Table","TV","AC","Cot"}   #Here the total home_appliances set is completely present in the second set so the second set is a superset
home_appliances_1 = {"Chair","Table","Frige","Sofa","TV","AC","Cot"}
home_appliances_1.issuperset(home_appliances)
```

* Removes common elements
* keeps non-common elements
* Returns a new set(symmetric_difference)
* Updaates it to a original set(symmetric_difference_update)

### 4. Dictionary
It is called as a Key:Value Pair

* Dict is Two dimensional (2D)
* It is heterogeneous
* Indexing and Slicing is not possible
* Don't create duplicate key , But you can create duplicate values

```python
home_appliances = {}
home_appliances
```

```python
type(home_appliances)
```

```python
home_appliances = {"TV":35000 , "AC":45000, "Washing machine":50000, "Fan":1500}
home_appliances
```

#### 1. clear - Remove all items from Dict

```python
home_appliances = {"TV":35000 , "AC":45000, "Washing machine":50000, "Fan":1500}
home_appliances.clear()
home_appliances
```

#### 2. copy - A shallow copy of Dict

```python
home_appliances = {"TV":35000 , "AC":45000, "Washing machine":50000, "Fan":1500}
home_appliances.copy()
home_appliances
```

#### 3. fromkeys - Create a new dictionary with given keys and a same value assigned to all keys

```python
home_appliances = {"TV", "AC", "Washing machine", "Fan"}
home = dict.fromkeys(home_appliances,90000)
home
```

```python
home_appliances = {"TV":35000 , "AC":45000, "Washing machine":50000, "Fan":1500}  #Here already values are assigned eventhough the values are reassigned to all values using this function
home_appliances.fromkeys(home_appliances,90000)
```

#### 4. get - Return the value for key if key is in the dictionary, else default.

```python
home_appliances = {"TV":35000 , "AC":45000, "Washing machine":50000, "Fan":1500}
home_appliances.get("Fan")
```

```python
home_appliances = {"TV":35000 , "AC":[90000,45000], "Washing machine":50000, "Fan":1500}
home_appliances.get("AC")
```
