#### 3.  If the user guessed number > secret number, then program should say. "Your number is lesser than the secret number"

```python
print("Welcome to the Number Guessing Game")
print("Guess the number between 1-100")

secret_number=98

for i in range(0,7):
    user_input= int(input("Enter your guess: "))

    if user_input < 1 or user_input > 100:
        print("Invalid number! Your number is must between in 1-100")
    elif user_input < secret_number:
        print("Your number is lesser")
    elif user_input > secret_number:
        print("Your number is higher")
    else:
        print("Your guessed number is correct")
        break
    
```

## Level 5: The secret number should not be fixed. It should be random.

#### There is an Library called random which we can generate the random numbers everytime.

#### 5.1 Using for loop

```python
import random

print("Welcome to the Number Guessing Game")
print("Guess the number between 1-100")

secret_number=random.randint(1,100)

for i in range(0,7):
    user_input= int(input("Enter your guess: "))

    if user_input < 1 or user_input > 100:
        print("Invalid number! Your number is must between in 1-100")
    elif user_input < secret_number:
        print("Your number is lesser")
    elif user_input > secret_number:
        print("Your number is higher")
    else:
        print("Your guessed number is correct")
        break


if secret_number == user_input:
    print(f"Congratulations ! You have guessed the number {secret_number} in {i} attempts")
else:
    print("Sorry ! All your trials are exceeded. Try Again")
```

## level 6: How to handle Invalid inputs like char or string or special char.

```python
import random

print("Welcome to the Number Guessing Game")
print("Guess the number between 1-100")

secret_number=random.randint(1,100)

for i in range(0,7):
    try:
        user_input= int(input("Enter your guess: "))
    
        if user_input < 1 or user_input > 100:
            print("Invalid number! Your number is must between in 1-100")
        elif user_input < secret_number:
            print("Your number is lesser")
        elif user_input > secret_number:
            print("Your number is higher")
        else:
            print("Your guessed number is correct")
            break
    except ValueError:
        print("Invalid characters, Not recognised. Please enter the number between 1-100")

if secret_number == user_input:
    print(f"Congratulations ! You have guessed the number {secret_number} in {i} attempts")
else:
    print("Sorry ! All your trials are exceeded. Try Again")
```
