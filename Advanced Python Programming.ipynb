## 5.4 Challenge 2 - Find the Factors of a Number
If the user enters 20 , output should be "The factors of 20 is 1,2,4,5,10,20"

```python
#user_input = 20
for i in range(1,25):
    if 20%i == 0:
        print(i)
   
```

```python
user_input = int(input("Enter your number"))

for i in range(1,user_input+1):
    if user_input % i == 0:
        print(i)
```

```python
user_input = int(input("Enter your number"))

result = []

for i in range(1,user_input+1):
    if user_input % i == 0:
        result.append(i)
        
print("The Factors of {} is {}".format(user_input,result))
```

## 5.5 Break out of a Pattern

```python
while True:
    exit_code = input("Enter the secret code to exit this window")
    if exit_code == 'q':
        break
```

```python
while True:
    exit_code = input("Enter the secret code to exit this window")
    if exit_code.lower() == 'q':
        break
```
