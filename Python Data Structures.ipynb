# 2. Sets

* Set is One dimensional (1D)
* It is heterogeneous
* Set won't allow duplicate values
* Indexing and Slicing is not possible
* Set is mutable

```python
empty_cart_2 = set()
```

```python
empty_cart_2
```

```python
type(empty_cart_2)
```

```python
items_to_buy_2 = {"Apple","Banana","Mango","Grapes","Apple", 8, 9.54, 10}
items_to_buy_2
```

```python
type(items_to_buy_2)
```

```python
items_to_buy_2[3]
```

```python
items_to_buy_2[0:5]
```

```python
items_to_buy_2.add("sapota")
items_to_buy_2
```

```python
items_to_buy_2.discard("Apple")
items_to_buy_2
```
